```python
import numpy as np
import matplotlib.pyplot as plt

# Generate simple data: y = 3x + 2 + noise
np.random.seed(0)
X = np.linspace(0, 10, 100)
y = 3 * X + 2 + np.random.randn(100)

# Gradient Descent
def gradient_descent(X, y, lr=0.01, epochs=100):
    m, b = 0, 0
    n = len(X)
    errors = []
    for _ in range(epochs):
        y_pred = m * X + b
        m -= lr * (-2/n) * np.sum(X * (y - y_pred))
        b -= lr * (-2/n) * np.sum(y - y_pred)
        errors.append(np.mean((y - y_pred)**2))
    return m, b, errors

# Stochastic Gradient Descent
def stochastic_gradient_descent(X, y, lr=0.01, epochs=20):
    m, b = 0, 0
    n = len(X)
    errors = []
    for _ in range(epochs):
        for i in range(n):
            xi, yi = X[i], y[i]
            y_pred = m * xi + b
            m -= lr * (-2) * xi * (yi - y_pred)
            b -= lr * (-2) * (yi - y_pred)
        y_pred_all = m * X + b
        errors.append(np.mean((y - y_pred_all)**2))
    return m, b, errors

# Run both
m_gd, b_gd, e_gd = gradient_descent(X, y)
m_sgd, b_sgd, e_sgd = stochastic_gradient_descent(X, y)

# Print results
print(f"GD:  m={m_gd:.2f}, b={b_gd:.2f}")
print(f"SGD: m={m_sgd:.2f}, b={b_sgd:.2f}")

# Plot comparison
plt.plot(e_gd, label="Gradient Descent")
plt.plot(e_sgd, label="Stochastic Gradient Descent")
plt.xlabel("Epochs")
plt.ylabel("Mean Squared Error")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate synthetic linear data: y = 3x + 2 + noise
np.random.seed(42)
X = np.linspace(0, 10, 100)
y = 3 * X + 2 + np.random.randn(100)

# Function to compute the mean squared error
def mse(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)

# Gradient Descent implementation
def gradient_descent(X, y, lr=0.01, epochs=1000):
    m, b = 0, 0
    n = len(X)
    errors = []

    for epoch in range(epochs):
        y_pred = m * X + b
        dm = (-2/n) * np.sum(X * (y - y_pred))
        db = (-2/n) * np.sum(y - y_pred)
        m -= lr * dm
        b -= lr * db
        errors.append(mse(y, y_pred))

    return m, b, errors

# Stochastic Gradient Descent implementation
def stochastic_gradient_descent(X, y, lr=0.01, epochs=100):
    m, b = 0, 0
    n = len(X)
    errors = []

    for epoch in range(epochs):
        total_error = 0
        for i in range(n):
            xi, yi = X[i], y[i]
            y_pred = m * xi + b
            dm = -2 * xi * (yi - y_pred)
            db = -2 * (yi - y_pred)
            m -= lr * dm
            b -= lr * db
            total_error += (yi - y_pred) ** 2
        errors.append(total_error / n)

    return m, b, errors

# Run both algorithms
m_gd, b_gd, err_gd = gradient_descent(X, y, lr=0.01, epochs=200)
m_sgd, b_sgd, err_sgd = stochastic_gradient_descent(X, y, lr=0.01, epochs=50)

# Print results
print(f"Gradient Descent: m = {m_gd:.3f}, b = {b_gd:.3f}")
print(f"Stochastic Gradient Descent: m = {m_sgd:.3f}, b = {b_sgd:.3f}")

# Plot the learning curves
plt.figure(figsize=(10,5))
plt.plot(err_gd, label="Gradient Descent")
plt.plot(err_sgd, label="Stochastic Gradient Descent")
plt.xlabel("Epochs")
plt.ylabel("Mean Squared Error")
plt.title("Convergence Comparison: GD vs SGD")
plt.legend()
plt.show()

# Plot regression lines
plt.figure(figsize=(10,5))
plt.scatter(X, y, color='blue', label='Data')
plt.plot(X, m_gd * X + b_gd, color='red', label='GD Line')
plt.plot(X, m_sgd * X + b_sgd, color='green', linestyle='--', label='SGD Line')
plt.legend()
plt.title("Linear Regression Fit: GD vs SGD")
plt.show()
```

```python
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense

# 1. Small training corpus
text = "I love machine learning. I love deep learning. I love natural language processing."

# 2. Tokenize words
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])
word_index = tokenizer.word_index
total_words = len(word_index) + 1

# 3. Create training sequences
input_sequences = []
for line in text.split('.'):
    tokens = tokenizer.texts_to_sequences([line])[0]
    for i in range(1, len(tokens)):
        input_sequences.append(tokens[:i+1])

# Pad sequences to equal length
max_seq_len = max([len(x) for x in input_sequences])
input_sequences = np.array(pad_sequences(input_sequences, maxlen=max_seq_len, padding='pre'))

# Split into inputs and labels
X = input_sequences[:, :-1]
y = input_sequences[:, -1]

# 4. Build simple RNN model
model = Sequential([
    Embedding(total_words, 10, input_length=max_seq_len-1),
    SimpleRNN(50),
    Dense(total_words, activation='softmax')
])

# 5. Compile and train
model.compile(loss='sparse_categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
model.fit(X, y, epochs=500, verbose=0)

# 6. Predict next word
def predict_next_word(seed_text):
    tokens = tokenizer.texts_to_sequences([seed_text])[0]
    tokens = pad_sequences([tokens], maxlen=max_seq_len-1, padding='pre')
    predicted = np.argmax(model.predict(tokens, verbose=0))
    for word, index in word_index.items():
        if index == predicted:
            return word

# 7. Try predictions
print("Next word prediction examples:")
for seed in ["I love", "I love deep", "machine"]:
    print(f"{seed} → {predict_next_word(seed)}")
```

```python
# Program 2: Load image dataset, normalize, and build simple ANN for image classification
# Author: Lokesh
# Subject: Deep Learning Lab

import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.utils import to_categorical

# 1. Load image dataset from Keras
(X_train, y_train), (X_test, y_test) = mnist.load_data()

# 2. Plot some images
plt.figure(figsize=(6, 3))
for i in range(6):
    plt.subplot(2, 3, i+1)
    plt.imshow(X_train[i], cmap='gray')
    plt.title(f"Label: {y_train[i]}")
    plt.axis('off')
plt.show()

# 3. Normalize the training and test data (0–1 range)
X_train = X_train / 255.0
X_test = X_test / 255.0

# 4. Convert labels to one-hot encoding
y_train = to_categorical(y_train)
y_test = to_categorical(y_test)

# 5. Build a simple Artificial Neural Network (ANN)
model = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation='relu'),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')  # 10 classes for digits 0–9
])

# 6. Compile and train the model
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
history = model.fit(X_train, y_train, epochs=5, batch_size=128, validation_split=0.1)

# 7. Evaluate model performance
test_loss, test_acc = model.evaluate(X_test, y_test)
print(f"Test Accuracy: {test_acc:.4f}")

# 8. Plot training history
plt.plot(history.history['accuracy'], label='Train Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.title('Training vs Validation Accuracy')
plt.legend()
plt.show()
```

```python
# Program 14: Sentiment Analysis using LSTM
# Author: Lokesh
# Subject: Deep Learning Lab

import numpy as np
from tensorflow.keras.datasets import imdb
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
from tensorflow.keras.preprocessing.sequence import pad_sequences
import matplotlib.pyplot as plt

# 1. Load IMDB dataset (top 10,000 most frequent words)
max_words = 10000
(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=max_words)

# 2. Pad sequences to have same length
max_len = 200
X_train = pad_sequences(X_train, maxlen=max_len)
X_test = pad_sequences(X_test, maxlen=max_len)

# 3. Build LSTM model
model = Sequential([
    Embedding(max_words, 128, input_length=max_len),
    LSTM(128, dropout=0.2, recurrent_dropout=0.2),
    Dense(1, activation='sigmoid')
])

# 4. Compile the model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# 5. Train the model
history = model.fit(X_train, y_train, epochs=3, batch_size=64, validation_split=0.2)

# 6. Evaluate on test data
test_loss, test_acc = model.evaluate(X_test, y_test)
print(f"Test Accuracy: {test_acc:.4f}")

# 7. Plot training vs validation accuracy
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.title('LSTM Sentiment Analysis Performance')
plt.legend()
plt.show()
```

```python
# Program 16: Flower Image Classification using ANN
# Author: Lokesh
# Subject: Deep Learning Lab

import tensorflow as tf
from tensorflow.keras import layers, models
import numpy as np
import matplotlib.pyplot as plt

# 1. Load the TensorFlow Flowers dataset
dataset_url = "https://storage.googleapis.com/download.tensorflow.org/example_images/flower_photos.tgz"
data_dir = tf.keras.utils.get_file('flower_photos', origin=dataset_url, untar=True)

# 2. Create training and validation sets
img_size = (180, 180)
batch_size = 32

train_ds = tf.keras.utils.image_dataset_from_directory(
    data_dir,
    validation_split=0.2,
    subset="training",
    seed=123,
    image_size=img_size,
    batch_size=batch_size
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    data_dir,
    validation_split=0.2,
    subset="validation",
    seed=123,
    image_size=img_size,
    batch_size=batch_size
)

# 3. Normalize pixel values
normalization_layer = layers.Rescaling(1./255)
train_ds = train_ds.map(lambda x, y: (normalization_layer(x), y))
val_ds = val_ds.map(lambda x, y: (normalization_layer(x), y))

# 4. Build a simple ANN model
model = models.Sequential([
    layers.Flatten(input_shape=(180, 180, 3)),
    layers.Dense(256, activation='relu'),
    layers.Dense(128, activation='relu'),
    layers.Dense(5, activation='softmax')  # 5 flower classes
])

# 5. Compile the model
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# 6. Train the model
history = model.fit(train_ds, validation_data=val_ds, epochs=5)

# 7. Plot training and validation accuracy
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.title('Flower Classification using ANN')
plt.legend()
plt.show()
```
